```
import numpy as np
import IPython.display as display
from matplotlib import pyplot as plt
import io
import base64

ys = 200 + np.random.randn(100)
x = [x for x in range(len(ys))]

fig = plt.figure(figsize=(4, 3), facecolor='w')
plt.plot(x, ys, '-')
plt.fill_between(x, ys, 195, where=(ys > 195), facecolor='g', alpha=0.6)
plt.title("Sample Visualization", fontsize=10)

data = io.BytesIO()
plt.savefig(data)
image = F"data:image/png;base64,{base64.b64encode(data.getvalue()).decode()}"
alt = "Sample Visualization"
display.display(display.Markdown(F"""![{alt}]({image})"""))
plt.close(fig)
```

```

import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt
import os
import zipfile
import glob
import re
from google.colab import files


print("Upload your Excel file and your shapefile ZIP folder.")
print("The ZIP should contain .shp, .shx, .dbf and .prj files.")

uploaded = files.upload()

# Extract shapefile ZIP if uploaded
for filename in uploaded:
    if filename.lower().endswith(".zip"):
        with zipfile.ZipFile(filename, "r") as zip_ref:
            zip_ref.extractall("/content/district_shp")

# Locate Excel file
excel_files = glob.glob("/content/*.xlsx")

if not excel_files:
    raise FileNotFoundError("Please upload your Excel file.")

excel_path = next(
    (f for f in excel_files if "cancer" in os.path.basename(f).lower()),
    excel_files[0]
)

# Locate shapefile
shp_files = glob.glob("/content/**/*.shp", recursive=True)

if not shp_files:
    raise FileNotFoundError(
        "No shapefile found. Upload a ZIP containing all shapefile components."
    )

shp_path = next(
    (f for f in shp_files if "district" in os.path.basename(f).lower()),
    shp_files[0]
)

print("\nExcel file:", excel_path)
print("Shapefile:", shp_path)

Cancer_incidents = pd.read_excel(
    excel_path,
    sheet_name="Long Format"
)

district = gpd.read_file(shp_path)

print("\nShapefile columns:")
print(district.columns.tolist())

print("\nExcel columns:")
print(Cancer_incidents.columns.tolist())


Cancer_incidents["District"] = (
    Cancer_incidents["District"].astype(str).str.strip()
)

Cancer_incidents["Year"] = pd.to_numeric(
    Cancer_incidents["Year"], errors="coerce"
)

Cancer_incidents["Estimated cases"] = pd.to_numeric(
    Cancer_incidents["Estimated cases"], errors="coerce"
)

Cancer_incidents = Cancer_incidents.dropna(
    subset=["District", "Year", "Estimated cases"]
)

# Convert long format into wide format
Cancer_incidents = Cancer_incidents.pivot_table(
    index="District",
    columns="Year",
    values="Estimated cases",
    aggfunc="sum"
).reset_index()

# Get year columns
years = sorted([
    col for col in Cancer_incidents.columns
    if isinstance(col, (int, float))
])

print("\nYears available:", years)


def clean_name(value):
    return re.sub(
        r"[^a-z0-9]",
        "",
        str(value).lower().strip()
    )

excel_districts = set(
    Cancer_incidents["District"].dropna().map(clean_name)
)

# Find the shapefile column with the most matching district names
best_column = None
best_matches = 0

for col in district.columns:
    if col == district.geometry.name:
        continue

    matches = set(
        district[col].dropna().astype(str).map(clean_name)
    )

    overlap = len(excel_districts.intersection(matches))

    if overlap > best_matches:
        best_matches = overlap
        best_column = col

if best_column is None or best_matches < 15:
    raise ValueError(
        "Could not identify the district-name column. "
        "Check the shapefile columns printed above."
    )

print("\nIdentified district column:", best_column)
print("Matching district names:", best_matches)


district["District_clean"] = (
    district[best_column].astype(str).map(clean_name)
)

Cancer_incidents["District_clean"] = (
    Cancer_incidents["District"].astype(str).map(clean_name)
)

district_map = district.merge(
    Cancer_incidents.drop(columns=["District"]),
    on="District_clean",
    how="left"
)

# Check missing districts
print("\nDistricts without data:")
print(
    district_map.loc[
        district_map[years].isna().all(axis=1),
        best_column
    ].tolist()
)


vmin = Cancer_incidents[years].min().min()
vmax = Cancer_incidents[years].max().max()


for year in years:

    fig, ax = plt.subplots(figsize=(10, 10))

    district_map.plot(
        column=year,
        cmap="BuPu",
        vmin=vmin,
        vmax=vmax,
        linewidth=0.6,
        edgecolor="black",
        legend=True,
        missing_kwds={
            "color": "lightgrey",
            "label": "No data"
        },
        legend_kwds={
            "label": "Estimated cancer cases",
            "shrink": 0.7
        },
        ax=ax
    )

    ax.set_title(
        f"Estimated Cancer Cases in Sri Lanka - {int(year)}",
        fontsize=15,
        fontweight="bold"
    )

    ax.axis("off")

    plt.tight_layout()
    plt.show()

print("\nAll yearly maps generated successfully!")
```

```

!pip -q install geopandas imageio

import os
import glob
import re
import numpy as np
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt
import imageio.v2 as imageio
from google.colab import files


excel_path = "/content/Cancer incidents 2000-2022.xlsx"
shp_path = "/content/Districts.shp"

if not os.path.exists(excel_path):
    matches = glob.glob("/content/**/*.xlsx", recursive=True)
    if not matches:
        raise FileNotFoundError("Cancer Excel file not found.")
    excel_path = next(
        (p for p in matches if "cancer" in os.path.basename(p).lower()),
        matches[0]
    )

if not os.path.exists(shp_path):
    matches = glob.glob("/content/**/*.shp", recursive=True)
    if not matches:
        raise FileNotFoundError("Districts.shp not found.")
    shp_path = next(
        (p for p in matches if "district" in os.path.basename(p).lower()),
        matches[0]
    )

print("Excel:", excel_path)
print("Shapefile:", shp_path)


pop_df = pd.read_excel(
    excel_path,
    sheet_name="Long Format"
)

pop_df.columns = [str(c).strip() for c in pop_df.columns]

# Clean and convert values
pop_df["District"] = pop_df["District"].astype(str).str.strip()
pop_df["Year"] = pd.to_numeric(pop_df["Year"], errors="coerce")
pop_df["Estimated cases"] = pd.to_numeric(
    pop_df["Estimated cases"].astype(str).str.replace(",", ""),
    errors="coerce"
)

pop_df = pop_df.dropna(
    subset=["District", "Year", "Estimated cases"]
)

# Convert long format into wide format
pop_df = pop_df.pivot_table(
    index="District",
    columns="Year",
    values="Estimated cases",
    aggfunc="sum"
).reset_index()

years = sorted([
    c for c in pop_df.columns
    if isinstance(c, (int, float))
])

if not years:
    raise ValueError("No yearly data found in the Excel sheet.")

print("Years:", years)


sl_shapefile = gpd.read_file(shp_path)

if sl_shapefile.crs is None:
    sl_shapefile = sl_shapefile.set_crs("EPSG:4326")

# Identify the correct district-name column automatically
def clean_name(value):
    return re.sub(r"[^a-z0-9]", "", str(value).lower())

excel_names = set(pop_df["District"].map(clean_name))

best_column = None
best_matches = 0

for col in sl_shapefile.columns:
    if col == sl_shapefile.geometry.name:
        continue

    shp_names = set(
        sl_shapefile[col].dropna().map(clean_name)
    )

    matches = len(excel_names.intersection(shp_names))

    if matches > best_matches:
        best_matches = matches
        best_column = col

if best_column is None or best_matches < 15:
    raise ValueError(
        "Could not identify district names. Check shapefile attributes."
    )

print("District-name column:", best_column)
print("Matched districts:", best_matches)


sl_shapefile["clean_dist"] = (
    sl_shapefile[best_column].map(clean_name)
)

pop_df["clean_dist"] = (
    pop_df["District"].map(clean_name)
)

merged_sl = sl_shapefile.merge(
    pop_df.drop(columns=["District"]),
    on="clean_dist",
    how="left"
)

# Check missing data
missing = merged_sl.loc[
    merged_sl[years].isna().all(axis=1),
    best_column
].tolist()

print("Districts without data:", missing)

if merged_sl[years].notna().sum().sum() == 0:
    raise ValueError("No cancer data matched the district shapefile.")


all_vals = pop_df[years].to_numpy(dtype=float)
all_vals = all_vals[np.isfinite(all_vals)]

vmin = np.min(all_vals)
vmax = np.max(all_vals)


os.makedirs("/content/temp_frames", exist_ok=True)

frame_paths = []

for year in years:

    fig, ax = plt.subplots(figsize=(8, 10))

    merged_sl.plot(
        column=year,
        cmap="OrRd",
        vmin=vmin,
        vmax=vmax,
        linewidth=0.5,
        edgecolor="black",
        missing_kwds={
            "color": "lightgrey",
            "label": "No data"
        },
        ax=ax
    )

    ax.set_title(
        f"Sri Lanka - Estimated Cancer Cases - {int(year)}",
        fontsize=14,
        fontweight="bold",
        pad=15
    )

    ax.set_axis_off()

    # Shared colour legend
    sm = plt.cm.ScalarMappable(
        cmap="OrRd",
        norm=plt.Normalize(vmin=vmin, vmax=vmax)
    )

    sm.set_array([])

    cbar = fig.colorbar(
        sm,
        ax=ax,
        shrink=0.65,
        pad=0.02
    )

    cbar.set_label(
        "Estimated Cancer Cases",
        fontsize=10
    )

    plt.tight_layout()

    frame_path = (
        f"/content/temp_frames/frame_{int(year)}.png"
    )

    # Keep identical image dimensions for every frame
    fig.savefig(frame_path, dpi=120)
    plt.close(fig)

    frame_paths.append(frame_path)

print("All yearly map frames created!")

images = [
    imageio.imread(path)
    for path in frame_paths
]

gif_path = "/content/Cancer_Incidents.gif"

imageio.mimsave(
    gif_path,
    images,
    duration=0.8,
    loop=0
)

print("Successfully created Cancer_Incidents.gif!")
print("Total frames:", len(images))

files.download(gif_path)
```
